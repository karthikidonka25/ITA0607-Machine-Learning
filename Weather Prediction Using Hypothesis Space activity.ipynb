```python
# ============================================
# Weather Prediction using Hypothesis Space
# Find-S Algorithm
# Google Colab Code
# ============================================

import pandas as pd
import numpy as np

# --------------------------------------
# Step 1: Create Dataset
# --------------------------------------

data = {
    'Outlook':['Sunny','Sunny','Overcast','Rainy','Rainy','Rainy','Overcast','Sunny'],
    'Temperature':['Hot','Hot','Hot','Mild','Cool','Cool','Mild','Mild'],
    'Humidity':['High','High','High','High','Normal','Normal','Normal','High'],
    'Wind':['Weak','Strong','Weak','Weak','Weak','Strong','Strong','Weak'],
    'PlayTennis':['No','No','Yes','Yes','Yes','No','Yes','No']
}

df = pd.DataFrame(data)

print("Dataset")
print(df)

# --------------------------------------
# Step 2: Separate Attributes & Target
# --------------------------------------

attributes = np.array(df.iloc[:, :-1])
target = np.array(df.iloc[:, -1])

# --------------------------------------
# Step 3: Find-S Algorithm
# --------------------------------------

specific_h = None

for i in range(len(target)):

    if target[i] == "Yes":

        if specific_h is None:
            specific_h = attributes[i].copy()

        else:
            for j in range(len(specific_h)):
                if specific_h[j] != attributes[i][j]:
                    specific_h[j] = '?'

        print("\nAfter Positive Example", i+1)
        print(specific_h)

# --------------------------------------
# Step 4: Final Hypothesis
# --------------------------------------

print("\n===================================")
print("Final Hypothesis")
print(specific_h)
print("===================================")

# --------------------------------------
# Step 5: Test New Weather Data
# --------------------------------------

new_weather = ['Overcast','Mild','Normal','Weak']

print("\nNew Weather Data")
print(new_weather)

prediction = "Yes"

for i in range(len(specific_h)):
    if specific_h[i] != '?' and specific_h[i] != new_weather[i]:
        prediction = "No"
        break

print("\nPrediction")
if prediction == "Yes":
    print("Suitable for Playing Tennis")
else:
    print("Not Suitable for Playing Tennis")
```
